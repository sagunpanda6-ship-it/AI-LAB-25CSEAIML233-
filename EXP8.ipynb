```python
from collections import deque
def water_jug(capacity1, capacity2, target):
  visited=set()
  queue=deque()
  queue.append((0,0,[]))
  while queue:
    jug1, jug2, path=queue.popleft()
    if (jug1, jug2) in visited:
      continue
    visited.add((jug1, jug2))
    path= path+ [(jug1, jug2)]
    if jug1 == target or jug2 == target:
      return path
    next_states=[
      (capacity1, jug2), # Fill jug1
      (jug1, capacity2), # Fill jug2
      (0, jug2),         # Empty jug1
      (jug1, 0),         # Empty jug2
      (
        jug1 - min(jug1, capacity2-jug2),
        jug2+min(jug1, capacity2-jug2)), # Pour from jug1 to jug2
        (jug1+min(jug2, capacity1-jug1), # Pour from jug2 to jug1
        jug2-min(jug2, capacity1-jug1))
    ]
    for state in next_states:
      if state not in visited:
        queue.append((state[0], state[1], path))
  return None
jug1=int(input("Enter the capacity of jug1: "))
jug2=int(input("Enter the capacity of jug2: "))
target=int(input("Enter the target volume: "))
solution=water_jug(jug1, jug2, target)
if solution is None:
  print("No solution found")
else:
  print("Solution path:")
  for state in solution:
    print(state)
```
