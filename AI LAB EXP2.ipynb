```python
def dfs(graph,node,visited):
  if node not in visited:
    print(node, end=" ")
    visited.add(node)

    for neighbor in graph[node]:
      dfs(graph, neighbor, visited)

graph={}

n=int(input("ENTER THE NYMBER OF VERTICES: "))

for i in range(n):
  vertex=input(f"enter the vertex {i+1}: ")
  neighbors= input(f"Enter adjacent vertices of{vertex}(space-separated): ").split()
  graph[vertex]= neighbors
start=input("enter the starting  index: ")
print("\ngraph:",graph)
print("DFS TRAVERASAL:", end=" ")

visited= set()
dfs(graph,start,visited)
```
