```python
from collections import deque
def bfs(graph,start):
  visited=set()
  queue=deque([start])
  visited.add(start)
  print("BFS TRAVERSAL:",end="")

  while queue:
    vertex=queue.popleft()
    print(vertex,end=" ")

    for neighbor in graph[vertex]:
      if neighbor not in visited:
        visited.add(neighbor)
        queue.append(neighbor)

graph={}

n=int(input("ENTER THE NUMBER OF VERTICES: "))


for i in range(n):

  vertex=input(f" ENTER VERTEX{i+1}:").upper()
  neighbors=[n.upper() for n in input(f"ENTER ADJACENT VERTICES OF {vertex}(space-separated):").split()]
  graph[vertex]=neighbors
start=input("ENTER THE STARTING VERTEX:").upper()


print("\ngraph:",graph)
bfs(graph,start)
```
